```python
import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
df = pd.read_csv('StudentsPerformance 1.csv')
```

Part 1: Verifying Assumptions (Pre-Analysis)

```python
n_total = len(df)
print(f"1. Sample size: {n_total}")
if n_total > 30:
    print("According to the Central Limit Theorem, for N > 30, the distribution of the sample mean")
    print("tends to be normal. Perfect normality of the data itself is not required.\n")
```

```python
# 2
female_scores = df[df['gender'] == 'female']['math score']
male_scores = df[df['gender'] == 'male']['math score']

f_success = (female_scores > 80).sum()
f_failure = (female_scores <= 80).sum()
m_success = (male_scores > 80).sum()
m_failure = (male_scores <= 80).sum()

print(f"Female: Success = {f_success}, failure = {f_failure}")
print(f"Male : Success = {m_success}, Failure = {m_failure}")

if all(x >= 10 for x in [f_success, f_failure, m_success, m_failure]):
    print("We can use z-test")
else:
    print("other test")
```

Part 2: Hypothesis Testing Tasks

```python
alpha = 0.05
print(" QUESTION 1")
mean_math = df['math score'].mean()
t_stat, p1 = stats.ttest_1samp(df['math score'], 65) 
print(t_stat)
print(f"maen math: {mean_math}")
print("p-value:", p1)

if p1 < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")

```

```python
print("\nQUESTION 2")

prep = df[df['test preparation course'] == 'completed']['math score']
none = df[df['test preparation course'] == 'none']['math score']

t_stat, p_one = stats.ttest_ind(
    prep,
    none,
    equal_var=False,           
    alternative='greater'      
)

print("p-value:", p_one)

if p_one < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
print("\nQUESTION 3")

mean_reading = df['reading score'].mean()
mean_writing = df['writing score'].mean()

t_stat, p3 = stats.ttest_rel(df['reading score'], df['writing score'])

print("p-value:", p3)

if p3 < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
print("\nQUESTION 4: Proportion Z-Test")

df['success'] = df['math score'] > 80

female = df[df['gender'] == 'female']
male = df[df['gender'] == 'male']

success_female = female['success'].sum()
success_male = male['success'].sum()

n_female = len(female)
n_male = len(male)

prop_female = success_female / n_female
prop_male = success_male / n_male

count = [success_female, success_male]
nobs = [n_female, n_male]
pooled_p = ((success_female+success_male)/(n_female+n_male))
stat, p4 = proportions_ztest(count, nobs)
print(pooled_p)
print("Proportion Female (Success):", round(prop_female, 3))
print("Proportion Male (Success):", round(prop_male, 3))
print("p-value:", p4)

if p4 < alpha:
    print("Decision: Reject the Null Hypothesis")
else:
    print("Decision: Fail to Reject the Null Hypothesis")
```
